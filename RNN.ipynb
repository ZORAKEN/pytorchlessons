An **RNN (Recurrent Neural Network)** is designed specifically for sequences—things where order matters.

📝 Words in a sentence

🎵 Notes in a song

📈 Stock/temperature measurements over time

🎙️ Audio/speech

🎥 Frames in a video

⌨️ Characters in text

At time step 1:

"I" → RNN → memory h₁

At time step 2:

"love" + h₁ → RNN → memory h₂

At time step 3:

"machine" + h₂ → RNN → memory h₃

So each step receives two things:

The current input
The previous memory

That's why it's called recurrent.

The basic RNN equation

You will probably see this equation when studying RNNs:

ht=tanh(Wx*xt+Wh*t−1+b)

```python
import numpy as np

# Input sequence
x = np.array([1, 2, 3, 4])

# RNN parameters
Wx = 0.5
Wh = 0.8
b = 0.0

# Initial hidden state
h = 0.0

# Process sequence
for value in x:
    h = np.tanh(Wx * value + Wh * h + b)
    print(h)
```

And then the new h replaces the old h.

That's the
**recurrence**

When using vectors.

For example:

input size = 3
hidden size = 4

Then:  x_t = [x1, x2, x3]   and:   h_t = [h1, h2, h3, h4]


Now the weights are matrices.

Wx → shape (hidden_size, input_size)

Wh → shape (hidden_size, hidden_size)

b → shape (hidden_size,)

h = np.tanh(Wx @ x + Wh @ h + b)

Here @ means matrix multiplication.

```python
import numpy as np

input_size = 3
hidden_size = 4

Wx = np.random.randn(hidden_size, input_size)
Wh = np.random.randn(hidden_size, hidden_size)
b = np.zeros(hidden_size)

h = np.zeros(hidden_size)

# Three time steps
x_sequence = [
    np.random.randn(input_size),
    np.random.randn(input_size),
    np.random.randn(input_size)
]

for x in x_sequence:

    h = np.tanh(
        Wx @ x
        + Wh @ h
        + b
    )

    print("Hidden state:", h)
```

***OUTPUT***

Now let's add an **output**
Suppose we want to classify the entire sequence.

We'll add:Wy

by

The output becomes: y=Wyh+by
	​


```python
import numpy as np

input_size = 3
hidden_size = 4
output_size = 2

Wx = np.random.randn(hidden_size, input_size)
Wh = np.random.randn(hidden_size, hidden_size)
b = np.zeros(hidden_size)

Wy = np.random.randn(output_size, hidden_size)
by = np.zeros(output_size)

h = np.zeros(hidden_size)

x_sequence = [
    np.random.randn(input_size),
    np.random.randn(input_size),
    np.random.randn(input_size)
]

for x in x_sequence:
    h = np.tanh(Wx @ x + Wh @ h + b)

# Final output
y = Wy @ h + by

print(y)
```

```python
import torch
import torch.nn as nn

class SimpleRNN(nn.Module):

    def __init__(self, input_size, hidden_size, output_size):
        super().__init__()

        self.rnn = nn.RNN(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True
        )

        self.fc = nn.Linear(
            hidden_size,
            output_size
        )

    def forward(self, x):

        output, hidden = self.rnn(x)

        # Take the last time step
        last_output = output[:, -1, :]

        # Classification
        prediction = self.fc(last_output)

        return prediction


model = SimpleRNN(
    input_size=10,
    hidden_size=20,
    output_size=2
)

x = torch.randn(32, 5, 10)

prediction = model(x)

print(prediction.shape)
```

prediction.shape is: torch.Size([32, 2])

PyTorch Training Pipeline

```python
#Step 1: Prepare the Dataset
X = torch.tensor([[1000.],
                  [1200.],
                  [1500.]])

y = torch.tensor([[200000.],
                  [250000.],
                  [310000.]])
```

```python
#Step 2: Create Dataset
from torch.utils.data import TensorDataset

dataset = TensorDataset(X, y)
"""
Index

0 → (1000,200000)

1 → (1200,250000)

2 → (1500,310000)"""
```

```python
#Step 3: DataLoader

#Training usually doesn't use the entire dataset at once.Instead it uses batches
from torch.utils.data import DataLoader

loader = DataLoader(dataset,
                    batch_size=32,
                    shuffle=True)
#Shufles mixes to avoid bias against a particular input
```

```python
#Step 4: Build the Model
import torch.nn as nn

model = nn.Linear(1,1)
```

```python
#Step 5: Loss Function
criterion = nn.MSELoss()
#if prediction=perfect ,loss =0
```

```python
#Step 6: Optimizer
optimizer = torch.optim.SGD(model.parameters(),
                            lr=0.01)
#move weights to reduce loss
```

```python
#Step 7: Forward Pass
predictions = model(X)
"""Input

↓

Model

↓

Prediction"""
```

```python
#Step 8: Compute Loss
loss = criterion(predictions,y)
```

```python
#Step 9: Backward Pass
#This is where Autograd works.

loss.backward()

#Autograd computes  dLoss/dWeight,dLoss/dBias
```

```python
#Step 10: Update Weights
optimizer.step()
#Current weight-(Gradient * Learning rate)
```
